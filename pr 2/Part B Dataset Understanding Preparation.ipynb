
# Part B: Dataset Understanding & Preparation

## Objectives
1. Identify features and target variable
2. Perform train-test split
3. Apply basic preprocessing (scaling where required)


```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv('Advanced Regression HousePrice.csv')

print("Dataset Shape:", df.shape)
df.head()
```

## Dataset Overview

```python
df.info()
```

```python
df.describe().T
```


## Task 1: Identify Features and Target Variable
Target Variable:
- house_price_inr

Features:
- area_sqft
- bedrooms
- bathrooms
- location_score
- property_age
- distance_city_km
- near_school
- near_metro
- crime_rate_index


```python
# Convert date column to datetime
df['sale_date'] = pd.to_datetime(df['sale_date'])

# Extract useful time features
df['sale_year'] = df['sale_date'].dt.year
df['sale_month'] = df['sale_date'].dt.month

# Feature Matrix
X = df.drop(columns=['house_price_inr', 'sale_date'])

# Target Variable
y = df['house_price_inr']

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

X.head()
```

## Task 2: Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)
print("Training Target   :", y_train.shape)
print("Testing Target    :", y_test.shape)
```


## Taskk 3: Feature Scaling

Scaling is applied because:
- Ridge Regression
- Lasso Regression
- Support Vector Regression

perform better when features are on a similar scale.


```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaled Training Shape:", X_train_scaled.shape)
print("Scaled Testing Shape :", X_test_scaled.shape)
```

```python
X_test_scaled
```

```python
scaled_df = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns
)

scaled_df.head()
```
